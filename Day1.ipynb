```python
#Subtract the Product and Sum of Digits of an Integer

n=int(input("Enter the number:"))
number=n
total_sum=0
prod=1
while number!=0:
    dig=number%10
    total_sum=total_sum+dig
    prod=prod*dig
    number=number//10
result=prod-total_sum
print("Total sum:",total_sum)
print("product:",prod)
print("result(sum-prod)",result)

```
