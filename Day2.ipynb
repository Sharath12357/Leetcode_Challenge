```python
N=int(input("Enter your number:"))
for i in range(1,N+1):
    if i%3==0 and i%5==0:
        print("FIZZBUZZ")
    elif i%3==0:
        print("FIZZ")
    elif i%5==0:
        print("BUZZ")
    else:
        print(i)
```

```python
def fixxbuzz(n):
    result=[]
    for i in range(1,n+1):
        if i%3==0 and i%5==0:
            result.append("FizzBuzz")
        elif i%3==0:
            result.append("Fizz")
        elif i%5==0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result


        
```

```python
fixxbuzz(5)
```
